* `mba_p` is MBA percentage. This is the student's grade from their MBA.

* For students who were placed, `salary` gives the starting salary of their job. For students who were not placed, the value is null.

````python
def convert_gender(text_gender:str) -> int:
  """Takes a string for the gender, and returns 0 if the string is 'M', otherwise returns 1"""
  if text_gender == "M":
    return 0
  else:
    return 1

# apply the conversion to all values in the Sex column
placement_df["gender"] = placement_df["gender"].apply(convert_gender)

placement_df["gender"].value_counts()
````

# Training a Perceptron

Now, we will train a perceptron on this data. We'll use the perceptron from Programming Assignment 1, and I have provided the correct code for you here.

````python
def unit_step(x):
  if x < 0:
    return 0
  return 1

def accuracy_score(y1, y2):
  if len(y1) != len(y2):
    raise Exception("Dimensions don't match")
  num_matches = 0
  for i in range(len(y1)):
    if y1[i] == y2[i]:
      num_matches += 1
  return num_matches / len(y1)
````

````python
class Perceptron:
  def __init__(self, num_features):
    self.weights = [0] * num_features
    self.bias = 0

  def predict_sample(self, x):
    total = self.bias
    for i in range(len(x)):
      total += self.weights[i] * x[i]
    return unit_step(total)

  def predict(self, X):
    return [self.predict_sample(x) for x in X]

  def train(self, X_train, y_train, learning_rate=.001, iterations=200):
    for iter in range(iterations):
      y_hat = self.predict(X_train)
      accuracy = accuracy_score(y_hat, y_train)
      #print("Accuracy after", iter, "epochs:", accuracy)
      for j in range(len(X_train)):
        if y_hat[j] != y_train[j]:
          for i in range(len(self.weights)):
            self.weights[i] += learning_rate * (y_train[j] - y_hat[j]) * X_train[j][i]
          self.bias += learning_rate * (y_train[j] - y_hat[j])
````

  Compute accuracy score for the perceptron
  Print current accuracy
  For each data point (x_1,...,x_M), with prediction y_hat and true value y:
